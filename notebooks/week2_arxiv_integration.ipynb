## 2. PDF Download and Caching

Test PDF download functionality with caching:

````python
async def download_all_pdfs(papers):
    """Download every paper PDF, using the client's cache."""

    print(f"PDF Download & Caching: {len(papers)} papers")

    downloaded_paths = []
    failed_papers = []

    for index, paper in enumerate(papers, start=1):
        print(f"\n[{index}/{len(papers)}] {paper.arxiv_id}")
        print(f"Title: {paper.title[:60]}...")

        try:
            pdf_path = await arxiv_client.download_pdf(paper)

            if pdf_path and pdf_path.exists():
                size_mb = pdf_path.stat().st_size / (1024 * 1024)
                downloaded_paths.append(pdf_path)
                print(f"✓ PDF ready: {pdf_path.name} ({size_mb:.2f} MB)")
            else:
                failed_papers.append((paper, "No file returned"))
                print("✗ PDF download failed")

        except Exception as exc:
            failed_papers.append((paper, str(exc)))
            print(f"✗ PDF download error: {exc}")

    print("\nDownload summary")
    print(f"✓ Successful: {len(downloaded_paths)}")
    print(f"✗ Failed: {len(failed_papers)}")

    if failed_papers:
        print("\nFailed papers:")
        for paper, reason in failed_papers:
            print(f"- {paper.arxiv_id}: {reason}")

    return downloaded_paths, failed_papers


pdf_paths, failed_papers = await download_all_pdfs(date_papers)
````

## 3. Docling PDF Processing

Test PDF parsing with Docling for structured content extraction:

````python
from datetime import datetime, timezone
from pathlib import Path

from dateutil import parser as date_parser

from src.config import get_settings
from src.db.factory import make_database
from src.repositories.paper import PaperRepository
from src.schemas.arxiv.paper import PaperCreate
from src.services.pdf_parser.factory import make_pdf_parser_service


async def parse_and_save_pdfs(papers, pdf_paths=None):
    """Parse downloaded PDFs with Docling and save results to PostgreSQL."""

    settings = get_settings()
    pdf_parser = make_pdf_parser_service()
    cache_dir = Path(settings.arxiv.pdf_cache_dir)

    # Match each cached filename to its arXiv metadata.
    papers_by_filename = {
        f"{paper.arxiv_id.replace('/', '_')}.pdf": paper
        for paper in papers
    }

    if pdf_paths is None:
        pdf_paths = sorted(cache_dir.glob("*.pdf"))
    else:
        pdf_paths = [Path(path) for path in pdf_paths]

    print("PDF Parsing with Docling")
    print("=" * 50)
    print(f"Parser limit: {settings.pdf_parser.max_pages} pages")
    print(f"File limit: {settings.pdf_parser.max_file_size_mb} MB")
    print(f"PDF files: {len(pdf_paths)}")

    results = {
        "parsed": [],
        "failed": [],
        "unmatched": [],
    }

    database = make_database()

    try:
        with database.get_session() as session:
            repository = PaperRepository(session)

            for index, pdf_path in enumerate(pdf_paths, start=1):
                paper = papers_by_filename.get(pdf_path.name)

                print(f"\n[{index}/{len(pdf_paths)}] {pdf_path.name}")

                if paper is None:
                    results["unmatched"].append(str(pdf_path))
                    print("⚠ No matching arXiv metadata; skipping")
                    continue

                try:
                    pdf_content = await pdf_parser.parse_pdf(pdf_path)
                    serialized = pdf_content.model_dump(mode="json")

                    # Metadata should normally already exist, but create it when
                    # the parsing cell is run independently.
                    stored_paper = repository.get_by_arxiv_id(
                        paper.arxiv_id
                    )

                    if stored_paper is None:
                        stored_paper = repository.create(
                            PaperCreate(
                                arxiv_id=paper.arxiv_id,
                                title=paper.title,
                                authors=paper.authors,
                                abstract=paper.abstract,
                                categories=paper.categories,
                                published_date=date_parser.parse(
                                    paper.published_date
                                ),
                                pdf_url=paper.pdf_url,
                                pdf_processed=False,
                            )
                        )

                    stored_paper.raw_text = serialized["raw_text"]
                    stored_paper.sections = serialized["sections"]
                    stored_paper.references = serialized["references"]
                    stored_paper.parser_used = serialized["parser_used"]
                    stored_paper.parser_metadata = {
                        **serialized["metadata"],
                        "source_pdf": str(pdf_path),
                        "figures": serialized["figures"],
                        "tables": serialized["tables"],
                    }
                    stored_paper.pdf_processed = True
                    stored_paper.pdf_processing_date = datetime.now(
                        timezone.utc
                    )

                    session.add(stored_paper)
                    session.commit()

                    results["parsed"].append(paper.arxiv_id)

                    print("✓ Parsed and saved")
                    print(
                        f"  Sections: {len(pdf_content.sections)}"
                    )
                    print(
                        f"  References: {len(pdf_content.references)}"
                    )
                    print(
                        f"  Text: {len(pdf_content.raw_text):,} characters"
                    )
                    print(f"  Parser: {pdf_content.parser_used.value}")

                except Exception as exc:
                    session.rollback()
                    results["failed"].append(
                        {
                            "arxiv_id": paper.arxiv_id,
                            "pdf": str(pdf_path),
                            "error": str(exc),
                        }
                    )
                    print(f"✗ Parsing failed: {exc}")

    finally:
        database.teardown()

    print("\nParsing summary")
    print(f"✓ Parsed and saved: {len(results['parsed'])}")
    print(f"✗ Failed:           {len(results['failed'])}")
    print(f"⚠ Unmatched:        {len(results['unmatched'])}")

    return results
````
