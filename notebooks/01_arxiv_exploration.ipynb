```python
# Initialize the downloader
downloader = ArXivDownloader(data_dir="../data")
print("✓ Downloader initialized")
print(f"  Data directory: {downloader.data_dir}")
print(f"  Metadata file: {downloader.metadata_file}")
```

```python
# Example 1: Search for Machine Learning papers (last 30 days)
print("\n=== Searching for Machine Learning Papers ===")
ml_papers = downloader.search(
    query="cat:cs.LG",  # Category: Machine Learning
    max_results=10,
    sort_order="submittedDate",
    sort_direction="descending"
)

print(f"Found {len(ml_papers)} papers")
if ml_papers:
    paper = ml_papers[0]
    print(f"\nFirst result:")
    print(f"  Title: {paper['title']}")
    print(f"  ArXiv ID: {paper['arxiv_id']}")
    print(f"  Authors: {', '.join(paper['authors'][:2])}...")
    print(f"  Published: {paper['published'][:10]}")
```

```python
# Example 2: Search for Natural Language Processing papers
print("\n=== Searching for NLP Papers ===")
nlp_papers = downloader.search(
    query="cat:cs.CL",  # Category: Computation and Language
    max_results=10,
    sort_order="submittedDate",
    sort_direction="descending"
)

print(f"Found {len(nlp_papers)} papers")
for paper in nlp_papers[:3]:
    print(f"  - {paper['arxiv_id']}: {paper['title'][:60]}...")
```

```python
# Example 3: Download and store metadata for papers
print("\n=== Downloading Paper Metadata ===")
downloader.download_papers(ml_papers, download_pdfs=False)
downloader.download_papers(nlp_papers, download_pdfs=False)

print(f"\nTotal papers in database: {len(downloader.metadata['papers'])}")
print("\nRecent papers:")
for paper in downloader.metadata['papers'][-3:]:
    print(f"  - {paper['arxiv_id']}: {paper['title'][:60]}...")
```

```python
# Example 4: Get papers by category (recent papers)
print("\n=== Recent Papers by Category ===")
computer_vision = downloader.get_papers_by_category(
    category="cs.CV",  # Computer Vision
    days_back=7,  # Last 7 days
    max_results=5
)

print(f"Found {len(computer_vision)} recent Computer Vision papers")
for paper in computer_vision:
    print(f"  - {paper['arxiv_id']}: {paper['title'][:50]}...")
```

```python
# Example 5: Search with keyword query (e.g., RAG papers)
print("\n=== Keyword Search: 'Retrieval Augmented Generation' ===")
rag_papers = downloader.search(
    query="all:Retrieval Augmented Generation",
    max_results=5
)

print(f"Found {len(rag_papers)} papers")
for paper in rag_papers:
    print(f"  - {paper['arxiv_id']}: {paper['title']}")
```

```python
# Example 6: Export metadata to CSV and view summary
print("\n=== Exporting Metadata ===")
downloader.download_papers(computer_vision, download_pdfs=False)
downloader.export_metadata_csv(output_file="papers_metadata.csv")

print(f"Metadata exported to: {downloader.processed_dir / 'papers_metadata.csv'}")

# Show the JSON metadata summary
with open(downloader.metadata_file, 'r') as f:
    metadata = json.load(f)
    print(f"\nTotal papers in JSON metadata: {len(metadata['papers'])}")
    print(f"Metadata file: {downloader.metadata_file}")
```

```python
# Reference: Common ArXiv Categories
print("\n=== Common ArXiv Categories ===")
categories = {
    "cs.AI": "Artificial Intelligence",
    "cs.CL": "Computation and Language (NLP)",
    "cs.CV": "Computer Vision",
    "cs.LG": "Machine Learning",
    "cs.IR": "Information Retrieval",
    "stat.ML": "Statistics - Machine Learning",
    "math.ST": "Mathematics - Statistics",
}

for code, name in categories.items():
    print(f"  {code}: {name}")

print("\nFor more categories, visit: https://arxiv.org/category_taxonomy")
```
